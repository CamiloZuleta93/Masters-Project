```python
import numpy as np
import pandas as pd

# File paths
input_file = 'CDI.csv'  # Original file
output_file = 'CDIAdjusted.csv'  # Cleaned file

try:
    # Load dataset
    df = pd.read_csv(input_file, sep=';')

    # Strip column names (remove extra spaces)
    df.columns = df.columns.str.strip()

    # Ensure required columns exist
    required_columns = ['COUNTRY', 'PDI', 'IDV', 'MAS', 'UAI', 'LTO', 'IVR']
    missing_columns = [col for col in required_columns if col not in df.columns]
    
    if missing_columns:
        raise ValueError(f"❌ Missing columns in dataset: {missing_columns}")

    # Convert numeric columns (handle errors)
    for col in ['PDI', 'IDV', 'MAS', 'UAI', 'LTO', 'IVR']:
        df[col] = pd.to_numeric(df[col], errors='coerce')

    # Drop NaNs
    df = df.dropna(subset=['PDI', 'IDV', 'MAS', 'UAI', 'LTO', 'IVR'])

    # Compute variance (ensure no zero values)
    variance = df[['PDI', 'IDV', 'MAS', 'UAI', 'LTO', 'IVR']].var()
    if (variance == 0).any():
        raise ValueError("❌ One or more dimensions have zero variance.")

    # Create all possible pairs
    pairs = df.merge(df, how='cross', suffixes=('_1', '_2'))
    pairs = pairs[pairs['COUNTRY_1'] != pairs['COUNTRY_2']]

    # Calculate CDI
    def cultural_distance(row):
        return np.sum(((row[['PDI_1', 'IDV_1', 'MAS_1', 'UAI_1', 'LTO_1', 'IVR_1']].values - 
                         row[['PDI_2', 'IDV_2', 'MAS_2', 'UAI_2', 'LTO_2', 'IVR_2']].values)) ** 2 / variance.values)

    pairs['CDI'] = pairs.apply(cultural_distance, axis=1)
    pairs = pairs[['COUNTRY_1', 'COUNTRY_2', 'CDI']]

    # Save file
    pairs.to_csv(output_file, index=False)
    print(f"✅ Cultural Distance Matrix saved successfully as {output_file}")

except FileNotFoundError:
    print(f"❌ The file {input_file} was not found.")
except ValueError as ve:
    print(f"❌ Error: {ve}")
except Exception as e:
    print(f"❌ Unexpected error: {e}")
```

```python
import pandas as pd
import re
from unidecode import unidecode
import logging

# File paths
input_file = 'country_languages.csv'  # Original file
output_file = 'country_languages_similarity.csv'  # Output file

# Set up logging for better debugging
logging.basicConfig(level=logging.INFO, format='%(asctime)s - %(message)s')

def clean_and_tokenize(text):
    """
    Cleans and tokenizes language data:
    - Removes text inside parentheses
    - Converts to lowercase
    - Splits by commas, handling extra spaces
    - Removes non-alphabetic characters except commas
    - Normalizes accents using unidecode
    """
    if pd.isna(text) or not isinstance(text, str):
        return set()
    
    text = unidecode(text)
    text = re.sub(r'\(.*?\)', '', text)
    text = re.sub(r'[^a-zA-Z, ]+', '', text)
    tokenized = set(re.split(r'\s*,\s*|\s+', text.lower().strip()))
    return tokenized

def simple_language_similarity(set1, set2):
    """
    Return 1 if the two sets share at least one language, otherwise 0.
    """
    return int(bool(set1 & set2))

# Your code logic (replacing print statements with logging)
logging.info(f"Processing file: {input_file}")

try:
    df = pd.read_csv(input_file, sep=';', encoding='utf-8', on_bad_lines='skip', quotechar='"')
    df.columns = df.columns.str.strip()

    required_columns = {'Country', 'Languages Spoken'}
    if not required_columns.issubset(df.columns):
        raise ValueError(f"❌ Missing required columns. Expected: {required_columns}, Found: {df.columns}")

    df = df.dropna(subset=['Country', 'Languages Spoken'])
    df['Country'] = df['Country'].apply(lambda x: unidecode(str(x)).strip().lower().capitalize())
    df['Country'] = df['Country'].replace({'Burkina faso': 'Burkina Faso'})  # Ensure proper capitalization
    
    countries = df['Country']
    languages = df['Languages Spoken']
    similarity_scores = []

    logging.info("Starting similarity calculations...")

    for i in range(len(countries)):
        for j in range(i + 1, len(countries)):
            country_1, country_2 = countries.iloc[i], countries.iloc[j]
            languages_1, languages_2 = clean_and_tokenize(languages.iloc[i]), clean_and_tokenize(languages.iloc[j])
            similarity_score = simple_language_similarity(languages_1, languages_2)
            similarity_scores.append([country_1, country_2, similarity_score])

    similarity_df = pd.DataFrame(similarity_scores, columns=['country_1', 'country_2', 'similarity_score'])
    similarity_df.to_csv(output_file, index=False, sep=';')

    logging.info(f"✅ Successfully generated the file: {output_file}")

except FileNotFoundError:
    logging.error(f"❌ The file {input_file} was not found.")
except ValueError as ve:
    logging.error(f"❌ Error: {ve}")
except Exception as e:
    logging.error(f"❌ Unexpected error: {e}")
```

```python
import pandas as pd
from thefuzz import process, fuzz

# Load los datos
archivo = "Homologar.xlsx"
df_transfer = pd.read_excel(archivo, sheet_name="Transfermarket")
df_fifa = pd.read_excel(archivo, sheet_name="FIFA")

# Lista de nombres únicos en Transfermarket
nombres_transfer = list(set(df_transfer["name"].tolist()))

# Conjunto para rastrear nombres ya asignados
nombres_usados = set()

# Función para homologar nombres asegurando unicidad
def homologar_nombre(nombre):
    global nombres_usados

    # Buscar las mejores coincidencias
    coincidencias = process.extract(nombre, nombres_transfer, scorer=fuzz.token_set_ratio, limit=5)

    # Filtrar solo coincidencias por encima del umbral de 75%
    mejores_opciones = [c[0] for c in coincidencias if c[1] >= 75 and c[0] not in nombres_usados]

    if not mejores_opciones:
        print(f"⚠ No se encontró un buen match para: {nombre}")
        return nombre  # Mantener original si no hay coincidencia aceptable

    # Tomar la mejor opción disponible
    mejor_match = mejores_opciones[0]
    nombres_usados.add(mejor_match)  # Marcar como usado
    return mejor_match

# Aplicar la homologación asegurando unicidad
df_fifa["Nombre Homologado"] = df_fifa["name"].apply(homologar_nombre)

# Guardar cambios en el archivo
with pd.ExcelWriter(archivo, engine="openpyxl", mode="a", if_sheet_exists="replace") as writer:
    df_fifa.to_excel(writer, sheet_name="FIFA", index=False)

print("✅ Proceso completado. Revisa la hoja 'FIFA' en el archivo Excel.")
```
